Customer analysis findings

1. Members are the stronger customer segment

Members generated 189,694.76 in sales.
Normal customers generated 133,271.99.
Members made 565 transactions, compared with 435 for Normal customers.
Average member transaction: 335.74
Average normal-customer transaction: 306.37

So members not only transact more frequently, but their average transaction is also higher.

2. Female customers generated substantially more sales

Female sales: 194,671.84
Male sales: 128,294.91
Female average transaction: 340.93
Male average transaction: 299.06

This is worth highlighting in your portfolio, but we should avoid claiming why females spend more unless the data provides evidence for the reason.

3. Cash generated the highest sales

Cash: 112,206.57
E-wallet: 109,993.11
Credit card: 100,767.07

Interestingly, E-wallet had the most transactions (345), while Cash generated the most sales. That's a good example of why transaction count and revenue should be analysed separately.
